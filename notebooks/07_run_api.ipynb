```python
# ========================================
# START AFYAMETRIX API — JUPYTER COMPATIBLE
# ========================================
# Jupyter runs its own async event loop, so we run
# uvicorn in a background thread to avoid the conflict.

import uvicorn
import threading
import sys
import time

sys.path.insert(0, r"C:\Users\hassa\afyametrix\api")

def run_server():
    uvicorn.run(
        "main:app",
        host="127.0.0.1",
        port=8000,
        reload=False,
        log_level="info"
    )

# Start server in background thread
# daemon=True means the thread dies when Jupyter shuts down
thread = threading.Thread(target=run_server, daemon=True)
thread.start()

# Give it 3 seconds to start up
time.sleep(3)

# Verify it's running
import requests
try:
    response = requests.get("http://127.0.0.1:8000/")
    data = response.json()
    print("✅ AfyaMetrix API is running!\n")
    print(f"   Status:   {data['status']}")
    print(f"   Version:  {data['version']}")
    print(f"\n📖 Interactive docs: http://127.0.0.1:8000/docs")
    print(f"🔗 Share with frontend team: http://127.0.0.1:8000/docs")
    print(f"\n📊 Datasets loaded:")
    for name, count in data['datasets_loaded'].items():
        print(f"   {name:<20} {count:,} rows")
except Exception as e:
    print(f"❌ Server didn't start: {e}")
    print("   Check the output above for error messages")
```
